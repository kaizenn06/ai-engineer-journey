```python
x = np.array([[[0, 1, 2, 3],
               [4, 5, 6, 7]],
              [[0, 1, 2, 3],
               [4, 5, 6, 7]],
              [[0, 1, 2, 3],
               [4, 5, 6, 7]]])
print(x.ndim) # number of axes, or dimension, of the array
print(x.size) # total number of elements of the array
print(x.shape) # the size of the array along each dimension
print(x.dtype) # describing the type of the elements in arrays
print(x.itemsize) # the size in bytes of each element of the array
```

```python
# [a11, a12, a13]
# [a21, a22, a23]
# [a31, a32, a33]
# Row-major:    a11 -> a12 -> a13 -> a21 -> a22 -> a23 -> a31 -> a32 -> a33
# Column-major: a11 -> a21 -> a31 -> a12 -> a22 -> a32 -> a13 -> a23 -> a33
array = np.array([[8, 6, 0, 9],
                  [7, 8, 6, 1],
                  [4, 1, 8, 2]])
```

```python
array.ravel() # Row-major
```

```python
array.ravel(order = 'F') # Column-major
```

```python
# reshape : Gives a new shape to an array without changing its data 
# transpose() : reverse or change the axes of an array according to the values you specify same transpose matrix
array = np.array([[8, 6, 0, 9],
                  [7, 8, 6, 1],
                  [4, 1, 8, 2]])
```

```python
array.reshape(6, 2) # order = 'C' : Row-major
```

```python
array.reshape((6, 2), order = 'F') # Column-major
```

```python
array.T # array.transpose()
```
