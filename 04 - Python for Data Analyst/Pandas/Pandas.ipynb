```python
'''
Filter a DataFrame
1. Single condition Filtering
    - Extracting a subset of data is perhaps the most common operation in data analysis. 
    A subset is a portion of a larger data set that fits some kind of condition
2. Multiple conditions Filtering
 a. Conditions
    AND &
    OR |
    NOT ~
 b. Equivalent methods
    ==         .eq
    !=         .ne
    <          .lt
    <=         .le
    >          .gt
    >=         .ge
 c. Methods
    .isin()
    .between()
    .isnull() : return True if NaN else False
    .notnull() : return False if NaN else True
'''
```

```python
df = pd.read_csv("data_filter.csv")
df
```

```python
df[df["EducationField"] == "Medical"]
```

```python
df["EducationField"] == "Medical"
```

```python
df[df["Gender"] == "Female"]
```

```python
df[df["Age"] > 40]
```

```python
df[(df["Gender"] == "Female") & (df["EducationField"] == "Medical")]
```

```python
df[(df["Gender"] == "Female") & (df["EducationField"] == "Medical") & (df["Age"] > 40)]
```

```python
df[(df["Gender"] == "Female") | (df["EducationField"] == "Medical")]
```

```python
df[~(df["Age"] > 40)]
```

```python
# df[(df["EducationField"] == "Medical") | (df["EducationField"] == "Life Sciences")]
df[df["EducationField"].isin(["Medical", "Life Sciences"])]
```

```python
# df[(df["Age"] >= 10) & (df["Age"] <= 50)]
df[df["Age"].between(10, 50)]
```
